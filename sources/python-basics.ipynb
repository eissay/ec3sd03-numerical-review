```python
# Import NumPy library
# NumPy is used for numerical computations and creating arrays of numbers
import numpy as np

# Import matplotlib library for plotting graphs
# pyplot provides functions similar to MATLAB plotting commands
import matplotlib.pyplot as plt


# -------------------------------------------------------
# Define the first mathematical function
# f(x) = x^2
# The function receives x as input and returns x squared
# -------------------------------------------------------
def f(x):
    return x**2


# -------------------------------------------------------
# Define the second mathematical function
# g(x) = 2x + 1
# This is a linear function
# -------------------------------------------------------
def g(x):
    return 2*x + 1


# -------------------------------------------------------
# Generate x values
# np.linspace(start, end, number_of_points)
# This creates 100 equally spaced values between -5 and 5
# More points make the graph smoother
# -------------------------------------------------------
x = np.linspace(-5, 5, 100)


# -------------------------------------------------------
# Evaluate the functions for all x values
# Because NumPy supports vector operations,
# the functions are applied to all elements of x
# -------------------------------------------------------
y1 = f(x)   # compute f(x) for each x
y2 = g(x)   # compute g(x) for each x


# -------------------------------------------------------
# Plot the first function
# x values on the horizontal axis
# y1 values on the vertical axis
# label is used later in the legend
# -------------------------------------------------------
plt.plot(x, y1, label="f(x) = x^2")


# -------------------------------------------------------
# Plot the second function on the same graph
# This allows us to visually compare the two functions
# -------------------------------------------------------
plt.plot(x, y2, label="g(x) = 2x + 1")


# -------------------------------------------------------
# Draw horizontal and vertical axes
# These help visualize where functions cross the axes
# -------------------------------------------------------
plt.axhline(0)   # x-axis
plt.axvline(0)   # y-axis


# -------------------------------------------------------
# Label the axes
# These describe what each axis represents
# -------------------------------------------------------
plt.xlabel("x")
plt.ylabel("y")


# -------------------------------------------------------
# Add a title to the graph
# -------------------------------------------------------
plt.title("Comparison of Two Functions")


# -------------------------------------------------------
# Display the legend
# The legend shows which curve corresponds to each function
# -------------------------------------------------------
plt.legend()


# -------------------------------------------------------
# Display the graph on the screen
# -------------------------------------------------------
plt.show()
```

# Code Reusability

A general plotting function allows you to reuse the same code for many different functions.

Instead of writing plotting code every time:

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt


# -------------------------------------------------------
# General plotting function
# This function receives:
# 1. func  -> the mathematical function to plot
# 2. x_min -> starting value of x
# 3. x_max -> ending value of x
# 4. n     -> number of points used for plotting
# -------------------------------------------------------
def plot_function(func, x_min, x_max, n=100):

    # Generate x values between x_min and x_max
    x = np.linspace(x_min, x_max, n)

    # Evaluate the function for all x values
    y = func(x)

    # Plot the function
    plt.plot(x, y)

    # Draw axes
    plt.axhline(0)
    plt.axvline(0)

    # Labels and title
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title("Function Plot")

    # Show the graph
    plt.show()
```

```python
# Example 1   
def f(x):
    return x**2

plot_function(f, -5, 5)
```

```python
# Example 2   

def g(x):
    return x**3 - x - 2

plot_function(g, -3, 3)
```

```python
plot_function(lambda x: x**3 - x - 2, -3, 3) # another way of defining a function 
```
